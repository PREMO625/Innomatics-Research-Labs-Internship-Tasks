## Task 1: Replace Hardcoded Prompts (10%)

**Given Code:**
```python
def explain_topic(topic):
   return f"Explain {topic} in simple terms for beginners"
```

**Task:**
Convert this into a LangChain `PromptTemplate`.

**Requirements:**
*   No f-strings
*   Use `PromptTemplate`
*   Must be reusable

````python
from langchain_core.prompts import PromptTemplate

# Original hardcoded function (for reference)
def explain_topic_hardcoded(topic):
   return f"Explain {topic} in simple terms for beginners"

# Create a PromptTemplate for Task 1
task1_template = PromptTemplate(
    template="Explain {topic} in simple terms for beginners",
    input_variables=["topic"]
)

# Demonstrate usage
print("--- Hardcoded Example ---")
print(explain_topic_hardcoded("Artificial Intelligence"))

print("\n--- PromptTemplate Example ---")
print(task1_template.format(topic="Artificial Intelligence"))

print("\n--- Another PromptTemplate Example ---")
print(task1_template.format(topic="Quantum Computing"))
````

## Task 4: ChatPromptTemplate System (15%)

**Build a chat-based prompt system using roles:**
*   `System` → defines behavior
*   `User` → dynamic input

**Roles:**
*   `teacher`
*   `interviewer`
*   `motivator`

**Example:**
`role = 'teacher'`
`topic = 'Neural Networks'`

````python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate

# Define a dictionary for system messages based on role
system_messages = {
    "teacher": "You are an experienced and patient educator. Your goal is to explain concepts clearly and help students understand the material thoroughly.",
    "interviewer": "You are a rigorous technical interviewer. Your goal is to assess the candidate's knowledge by asking challenging questions.",
    "motivator": "You are an inspiring motivational coach. Your goal is to encourage and uplift the user to achieve their full potential."
}

def create_chat_prompt_for_role(role, topic):
    if role not in system_messages:
        raise ValueError(f"Invalid role: {role}. Choose from {list(system_messages.keys())}")

    # Create a system message prompt template dynamically based on the role
    system_message_prompt = SystemMessagePromptTemplate.from_template(system_messages[role])

    # Create a human message prompt template for the topic
    human_message_prompt = HumanMessagePromptTemplate.from_template("Explain {topic} or ask questions about {topic}.")

    # Combine them into a ChatPromptTemplate
    chat_template = ChatPromptTemplate.from_messages([
        system_message_prompt,
        human_message_prompt
    ])
    return chat_template.format_messages(topic=topic)


# Test Cases
print("--- ChatPromptTemplate System Examples ---")

# Example 1: Teacher explaining Neural Networks
role_1 = 'teacher'
topic_1 = 'Neural Networks'
print(f"Role: {role_1}, Topic: {topic_1}")
chat_prompt_1 = create_chat_prompt_for_role(role_1, topic_1)
for msg in chat_prompt_1:
    print(f"{msg.type.capitalize()}: {msg.content}")
print("\n")

# Example 2: Interviewer asking about Reinforcement Learning
role_2 = 'interviewer'
topic_2 = 'Reinforcement Learning'
print(f"Role: {role_2}, Topic: {topic_2}")
chat_prompt_2 = create_chat_prompt_for_role(role_2, topic_2)
for msg in chat_prompt_2:
    print(f"{msg.type.capitalize()}: {msg.content}")
print("\n")

# Example 3: Motivator about Data Science career
role_3 = 'motivator'
topic_3 = 'Data Science career'
print(f"Role: {role_3}, Topic: {topic_3}")
chat_prompt_3 = create_chat_prompt_for_role(role_3, topic_3)
for msg in chat_prompt_3:
    print(f"{msg.type.capitalize()}: {msg.content}")
print("\n")
````

## Task 5: Input Validation Layer (10%)

**Create function:**
`validate_inputs(audience, tone)`

**Rules:**
*   `audience` ∈ `[beginner, intermediate, expert]`
*   `tone` ∈ `[formal, casual, fun]`

**If invalid:**
*   Raise error OR
*   Assign default values

````python
def validate_inputs(audience, tone, use_defaults=False):
    """
    Validates audience and tone inputs. Raises ValueError or assigns defaults.
    """
    valid_audiences = ['beginner', 'intermediate', 'expert']
    valid_tones = ['formal', 'casual', 'fun']

    validated_audience = audience
    validated_tone = tone

    if audience not in valid_audiences:
        if use_defaults:
            validated_audience = 'beginner' # Default audience
            print(f"Warning: Invalid audience '{audience}'. Using default: '{validated_audience}'.")
        else:
            raise ValueError(f"Invalid audience: '{audience}'. Must be one of {valid_audiences}.")

    if tone not in valid_tones:
        if use_defaults:
            validated_tone = 'formal' # Default tone
            print(f"Warning: Invalid tone '{tone}'. Using default: '{validated_tone}'.")
        else:
            raise ValueError(f"Invalid tone: '{tone}'. Must be one of {valid_tones}.")

    return validated_audience, validated_tone

print("--- Input Validation Layer Examples ---")

# Test valid inputs
try:
    aud, ton = validate_inputs('intermediate', 'casual')
    print(f"Valid inputs: audience='{aud}', tone='{ton}' (No error)\n")
except ValueError as e:
    print(f"Error: {e}\n")

# Test invalid audience, raise error
try:
    aud, ton = validate_inputs('advanced', 'formal')
    print(f"Valid inputs: audience='{aud}', tone='{ton}' (Should not print)\n")
except ValueError as e:
    print(f"Error with invalid audience (raise error): {e}\n")

# Test invalid tone, assign defaults
try:
    aud, ton = validate_inputs('expert', 'silly', use_defaults=True)
    print(f"Invalid tone, with defaults: audience='{aud}', tone='{ton}'\n")
except ValueError as e:
    print(f"Error: {e}\n")

# Test both invalid, assign defaults
try:
    aud, ton = validate_inputs('pro', 'happy', use_defaults=True)
    print(f"Both invalid, with defaults: audience='{aud}', tone='{ton}'\n")
except ValueError as e:
    print(f"Error: {e}\n")
````

## Task 6: Prompt Generator App (15%)

**Create function:**
`generate_prompt(topic, audience, tone, style)`

**Requirements:**
*   Use `PromptTemplate`
*   **Styles:** `teaching`, `interview`, `storytelling` (reusing templates from Task 3, but integrating with audience/tone)

**Example Output:**
`Explain Neural Networks for beginners in a fun storytelling style`

````python
def generate_prompt(topic, audience, tone, style, use_defaults_for_validation=False):
    """
    Generates a structured prompt based on topic, audience, tone, and a chosen style.
    Includes input validation for audience and tone.
    """
    # 1. Validate inputs (using the function from Task 5)
    try:
        validated_audience, validated_tone = validate_inputs(audience, tone, use_defaults=use_defaults_for_validation)
    except ValueError as e:
        print(f"Prompt generation failed due to invalid input: {e}")
        return None

    # 2. Define style-specific templates (combining ideas from Task 2 and Task 3)
    # We'll create a single template that can accommodate all inputs and styles
    # The core structure comes from the multi-input template, and 'style' modifies the explanation part.

    style_templates = {
        "teaching": "Explain {topic} clearly step by step for {audience} in a {tone} tone.",
        "interview": "Ask 3 insightful questions about {topic} for an {audience} level candidate in a {tone} manner.",
        "storytelling": "Explain {topic} as a captivating story for {audience} in a {tone} style."
    }

    if style not in style_templates:
        print(f"Invalid style: '{style}'. Choose from {list(style_templates.keys())}")
        return None

    full_prompt_template = PromptTemplate(
        template=style_templates[style],
        input_variables=["topic", "audience", "tone"]
    )

    # 3. Generate the prompt
    final_prompt = full_prompt_template.format(
        topic=topic,
        audience=validated_audience,
        tone=validated_tone
    )

    return final_prompt

print("--- Prompt Generator App Examples ---")

# Example 1: Neural Networks for beginners in a fun storytelling style
output_prompt_1 = generate_prompt(
    topic='Neural Networks',
    audience='beginner',
    tone='fun',
    style='storytelling'
)
if output_prompt_1:
    print(f"Generated Prompt 1: {output_prompt_1}\n")

# Example 2: Python for intermediate users in a formal teaching style
output_prompt_2 = generate_prompt(
    topic='Python',
    audience='intermediate',
    tone='formal',
    style='teaching'
)
if output_prompt_2:
    print(f"Generated Prompt 2: {output_prompt_2}\n")

# Example 3: Quantum Physics for experts in a casual interview style
output_prompt_3 = generate_prompt(
    topic='Quantum Physics',
    audience='expert',
    tone='casual',
    style='interview'
)
if output_prompt_3:
    print(f"Generated Prompt 3: {output_prompt_3}\n")

# Example 4: Invalid input with error handling (no defaults)
output_prompt_4 = generate_prompt(
    topic='Blockchain',
    audience='novice',
    tone='serious',
    style='teaching'
)
if output_prompt_4:
    print(f"Generated Prompt 4: {output_prompt_4}\n")

# Example 5: Invalid input with default assignment
output_prompt_5 = generate_prompt(
    topic='Cybersecurity',
    audience='newbie',
    tone='serious',
    style='storytelling',
    use_defaults_for_validation=True
)
if output_prompt_5:
    print(f"Generated Prompt 5 (with defaults): {output_prompt_5}\n")
````
